```python
# lamba functions
# IT IS A ONE LINE ANONYMOUS FUNCTION , it can accept any number of parameters and it returns a valid python expression.
addition_1= lambda x:x+1 # whatever is written after : is the expression that will be returned by lambda function
result_1=addition_1(1)
print(result_1)
```

```python
addition_2= lambda x ,y : x+y
result_2=addition_1(1,2)
print(result_2)
```

```python
def addition_2(x,y):
  return x+y
  '''i can use this function but what if i need to perform
  the addition once in the program then we can use lambda function
  it basically reduces the complexity.
  IT IS MAILY USED IN --> MAPS , FILTERS , SORTS'''
```

```python
#MAPS --> map function take some list / iterable object and apply the function to every single element inside that list
my_num=[1,2,3,4] # list

def square(x): # def a function to calculate square
  return x**2

square = list(map(square,my_num)) # use a list , map each item to square function
print(square)
```

```python
# MAPS WITH LAMBDA FUNCTION
my_num=[1,2,3,4] # list
''' insted of this -->>
def square(x): # def a function to calculate square
  return x**2 --> use lambda function
'''

square = list(map(lambda x : x**2 ,my_num)) # we had passed lambda function as an argument to another function
print(square)
```

```python
# FILTERS --> it applies the (lambda) function to every single iotem inside the list
#or iterable ojject and if the function return true then we keep the item
#else we reject the item.
even= list(filter(lambda x: x%2==0 ,my_num))
print(even)
```

```python
#SORT-->it sorts the items of a LIST
values=[(1,'b' ,"hello"),(2,'a',"world"),(3,'c',"okay")]
sorted_values=sorted(values,key=lambda x: x[2]) #we used lambda function to define the key [0]/[1]/[2]
print(sorted_values)
```

```python
#
values=[(1,'b' ,"hello"),(2,'a',"world"),(3,'c',"okay")]
sorted_values=sorted(values,key=lambda x: x[1]+x[2]) #we used lambda function to define the key [0]/[1]/[2]
print(sorted_values)
```

```python
num=[1,2,3,4,5]
from functools import reduce
sumofnumber= reduce(lambda acc, x: acc+x,num)
print(sumofnumber)
# reduce--> it is going to take list or an iterable object and reduce it to a single value
#accumulated(acc)--> by default it is always num[0]
#x is by default num[1]
#acc=1,x=2--> acc=x=3
#acc=3,x=3--> acc=x=6
#acc=6,x=4--> acc=x=10
#acc=10,x=5--> acc=x=15
#acc=15
```

```python
maxvalue = reduce(lambda acc , x : acc if acc > x else x, num)
print(maxvalue)
# accumulated(acc)--> by default it is always num[0]
# x is by default num[1]
# acc =1, x =2--> acc >x , x-> 2
# acc =2, x =3--> acc >x , x-> 3
# acc =3, x =4--> acc >x , x-> 4
# acc =4, x =5--> acc >x , x-> 5
# acc =5
```
